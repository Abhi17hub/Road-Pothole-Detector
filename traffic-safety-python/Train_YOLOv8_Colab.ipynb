# 🚀 Fine-Tuning YOLOv8 for Road Safety & Pothole Detection on Google Colab

This notebook trains YOLOv8 on custom road damage datasets (Potholes, Cracks, Damaged Surfaces) using a free GPU in Google Colab.

```python
# 1. Check GPU availability & install Ultralytics YOLOv8
!nvidia-smi
!pip install ultralytics roboflow
```

```python
# 2. Download Public Road Damage / Pothole Dataset from Roboflow
import roboflow

# Downloads public open-source Pothole Dataset in YOLOv8 format
!curl -L "https://universe.roboflow.com/ds/S69NfR8wKz?key=vT6x1R4yZ2" > pothole_dataset.zip
!unzip -q pothole_dataset.zip -d ./pothole_data
!rm pothole_dataset.zip
print("Dataset successfully extracted to ./pothole_data")
```

```python
# 3. Train YOLOv8 model on GPU
from ultralytics import YOLO

# Load pre-trained base model
model = YOLO('yolov8n.pt')

# Fine-tune on downloaded road dataset
results = model.train(
    data='./pothole_data/data.yaml',
    epochs=50,
    imgsz=640,
    batch=16,
    name='road_damage_yolov8',
    device=0
)
print("Training complete! Weights saved at runs/detect/road_damage_yolov8/weights/best.pt")
```

```python
# 4. Validate Model Performance
metrics = model.val()
print("mAP50 Accuracy:", metrics.box.map50)
```
